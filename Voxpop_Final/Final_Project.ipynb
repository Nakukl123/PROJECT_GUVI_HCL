```python
def clean_text(text):

    # remove urls
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # remove mentions
    text = re.sub(r'@\w+', '', text)

    # remove hashtags
    text = re.sub(r'#\w+', '', text)

    # remove special chars
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # tokenization
    tokens = text.split()

    # remove stopwords + lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
        if word not in stop_words
    ]

    return " ".join(tokens)
```

```python
class TweetDataset(Dataset):

    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):

        return (
            torch.tensor(
                self.X[idx],
                dtype=torch.long
            ),

            torch.tensor(
                self.y[idx],
                dtype=torch.float32
            )
        )
```

```python
train_dataset = TweetDataset(
    X_train,
    y_train
)

test_dataset = TweetDataset(
    X_test,
    y_test
)
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)
```

```python
for X_batch, y_batch in train_loader:

    print(X_batch.shape)
    print(y_batch.shape)

    break
```

```python
import torch.nn as nn

class AngerLSTM(nn.Module):

    def __init__(self, vocab_size):

        super().__init__()

        self.embedding = nn.Embedding(
            num_embeddings=vocab_size + 1,
            embedding_dim=64
        )

        self.lstm = nn.LSTM(
            input_size=64,
            hidden_size=64,
            batch_first=True
        )

        self.fc = nn.Linear(
            64,
            1
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        x = self.embedding(x)

        output, (hidden, cell) = self.lstm(x)

        x = hidden[-1]

        x = self.fc(x)

        x = self.sigmoid(x)

        return x.squeeze()
```

```python
vocab_size = len(word2idx)

model = AngerLSTM(vocab_size)

print(model)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=0.0001
)
```

```python
print(model)
```

```python
criterion = nn.BCELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=0.0001
)
```

```python
num_epochs = 10

for epoch in range(num_epochs):

    model.train()

    total_loss = 0

    for X_batch, y_batch in train_loader:

        optimizer.zero_grad()

        outputs = model(X_batch)

        loss = criterion(
            outputs,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    print(
        f"Epoch {epoch+1}/{num_epochs}, "
        f"Loss: {total_loss/len(train_loader):.6f}"
    )
```

```python
model.eval()

predictions = []
actuals = []

with torch.no_grad():

    for X_batch, y_batch in test_loader:

        outputs = model(X_batch)

        predictions.extend(
            outputs.numpy()
        )

        actuals.extend(
            y_batch.numpy()
        )
```

```python
import pandas as pd

results = pd.DataFrame({
    "Actual": actuals,
    "Predicted": predictions
})

results.head(20)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    results["Actual"],
    results["Predicted"]
)

print("MAE:", round(mae, 4))
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(
    results["Actual"],
    results["Predicted"]
)

print("MSE:", round(mse, 4))
```
